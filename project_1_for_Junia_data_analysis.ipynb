```python
import pandas as pd
```

```python
df = pd.read_excel("/content/Dataset for Data Analytics.xlsx")
```

```python
print("Dataset Shape (Rows, Columns):", df.shape)
print("\nFirst 5 Rows:")
display(df.head())
print("\nData Types & Info:")
print(df.info())
```

```python
print("Missing values per column before cleaning:")
print(df.isnull().sum())
```

```python
if 'CouponCode' in df.columns:
    df['CouponCode'] = df['CouponCode'].fillna('NONE')
    print("\nMissing values after imputation:", df['CouponCode'].isnull().sum())
```

```python
print("Duplicate OrderIDs found:", df['OrderID'].duplicated().sum())
print("Exact duplicate rows found:", df.duplicated().sum())
```

```python
df = df.drop_duplicates(subset=['OrderID'], keep='first')
```

```python
string_columns = df.select_dtypes(include=['object']).columns
```

```python
for col in string_columns:
    # Trim leading/trailing spaces and capitalize each word properly
    df[col] = df[col].astype(str).str.strip().str.title()
```

```python
df['CouponCode'] = df['CouponCode'].replace('None', 'NONE')
```

```python
# Convert Date to ISO 8601 standard (YYYY-MM-DD)
df['Date'] = pd.to_datetime(df['Date']).dt.strftime('%Y-%m-%d')
```

```python
# Round numerical price columns to 2 decimal places
df['UnitPrice'] = df['UnitPrice'].round(2)
df['TotalPrice'] = df['TotalPrice'].round(2)
```

```python
print(f"Total Missing Values Remaining: {df.isnull().sum().sum()}")
print(f"Duplicate OrderIDs Remaining: {df['OrderID'].duplicated().sum()}")
print(f"Final Cleaned Dataset Shape: {df.shape}")
```

```python
df.to_excel('Cleaned_Dataset_Data_Analytics.xlsx', index=False)
print("\nCleaned file successfully exported as 'Cleaned_Dataset_Data_Analytics.xlsx'!")
```
